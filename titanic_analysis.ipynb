# 鐵達尼號乘客生還分析
分析鐵達尼號乘客資料，探討性別、艙等與生還率的關係。

```python
import pandas as pd
import matplotlib.pyplot as plt

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
print(f"資料筆數：{df.shape[0]}，欄位數：{df.shape[1]}")
```

```python
print(f"平均年齡：{df['Age'].mean():.1f} 歲")
print(f"生還率：{df['Survived'].mean()*100:.1f}%")
print(f"生還：{df['Survived'].sum()} 人，罹難：{(df['Survived']==0).sum()} 人")
```

```python
df.groupby('Sex')['Survived'].mean().plot(
    kind='bar', color=['red','steelblue'],
    title='Survival Rate by Sex')
plt.ylabel('Survival Rate')
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python

艙等生還率=df.groupby('Pclass')['Survived'].mean()

艙等生還率.plot(
    kind="bar",
    title="Survival Rate by Passenger Class",
    color=["pink","yellow","green"]
    )
plt.xlabel("Passenger Class")
plt.ylabel("Survival Rate")
plt.ylim(0,1)
plt.xticks(rotation=0)
plt.xticks([0, 1, 2], ["1st Class", "2nd Class", "3rd Class"])
plt.tight_layout()
plt.show()
```

```python
df["Age"].plot(kind="hist", bins=20, color="grey")
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Number of people")
plt.show()
```

## 結論
1. 女性生還率（74%）遠高於男性（19%）
2. 艙等越高生還率越高（頭等63%，三等24%）
3. 乘客年齡集中在 20~40 歲
4. 船上有一定數量的幼童（0~5歲約40人）
